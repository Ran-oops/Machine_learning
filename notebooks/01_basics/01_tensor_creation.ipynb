```python
# 1.返回的x为未初始化的张量
x = torch.empty(5,3)
x

"""
torch.empty() 是一个函数，它可以创建一个未初始化的张量。这意味着张量中的值是随机的，没有被设置为任何特定的值1。
这个函数通常用于创建一个张量，然后再用其他值填充它。
0维数组--标量 1维数组---向量 2维数组， 三维数组  n维张量
"""
```

```python
# 2.创建一个5*3的随机初始化的张量
x=torch.rand(5,3)
x
"""
torch.empty() 和 torch.rand() 都可以用来创建张量，但它们的区别在于张量中的值。
torch.empty() 创建一个未初始化的张量，这意味着张量中的值是随机的，没有被设置为任何特定的值1。
而 torch.rand() 则创建一个随机初始化的张量，其中的值是从区间 [0,1) 的均匀分布中随机生成的1。
"""
```
